```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load directly from your GitHub repository (replace 'Pauloade123' if using a different handle)
url = "https://raw.githubusercontent.com/Pauloade123/ML-intern/main/data/raw/content_refresh_anonymized.csv"

import pandas as pd

# 1. Load the dataset from your personal repository
url = "https://raw.githubusercontent.com/Pauloade123/ML-intern/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)
df.columns

# 2. Compute the 3 supporting metrics
total_pages = len(df)
avg_ctr = df['ctr'].mean() * 100
opportunity_pages = len(df[(df['impressions_90d'] > df['impressions_90d'].median()) & (df['ctr'] < df['ctr'].mean())])

# 3. Print the results
print(f"Total pages: {total_pages:,}")
print(f"Average CTR: {avg_ctr:.2f}%")
print(f"High-impression / Low-CTR opportunity pages: {opportunity_pages:,}")

```

Across a total dataset of 30,000 pages, the overall average CTR is 51.07%. However, 12,866 pages have above-median 90-day impressions while falling below the average CTR, representing a massive pool of high-visibility pages that can yield quick traffic gains from title and metadata optimizations
